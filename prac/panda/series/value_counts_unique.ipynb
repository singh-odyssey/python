## Question 12: `value_counts()`, `unique()`, `nunique()`

### Tasks:
1. Create a Series of 10 city names where some cities **repeat**
2. Use `value_counts()` to find **how many times** each city appears
3. Use `unique()` to get the **list of unique cities** (no duplicates)
4. Use `nunique()` to find **how many unique cities** there are
5. Use `value_counts(normalize=True)` to get the **percentage share** of each city
6. Create a Series of exam results `['pass', 'fail', 'pass', 'pass', 'fail', 'pass']` and find what **percentage** passed vs failed

```python
import pandas as pd
```

```python
# Task 1: Create a Series of 10 city names with some repeats
cities =pd.Series(['Ahmedabad', 'Bangalore', 'Chennai', 'Surat', 'Hyderabad', 'Jaipur', 'Surat', 'Mumbai', 'Pune', 'Surat'])
```

```python
# Task 2: How many times does each city appear?
print(cities.value_counts())
```

```python
# Task 3: Get the list of unique cities (no duplicates)
print(cities.unique())
```

```python
# Task 4: How many unique cities are there?
print(cities.nunique())
```

```python
# Task 5: Percentage share of each city using normalize=True
print(cities.value_counts(normalize=True))
```

```python
# Task 6: Pass/fail percentage
result =pd.Series(['pass', 'fail', 'pass', 'pass', 'fail', 'pass'])
print(result.value_counts(normalize=True))
```
