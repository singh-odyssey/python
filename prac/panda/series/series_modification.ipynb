```python
import pandas as pd
```

```python
s= pd.Series([100,200,300,400,500],index=("pen","pen2","pen3","pen4","pen5"))
display(s)
```

```python
s["pen"]=101
display(s)
```

```python
s.iloc[0]=102
display(s)
```

```python
s['pen6']=106
s['pen7']=107
display(s)
```
