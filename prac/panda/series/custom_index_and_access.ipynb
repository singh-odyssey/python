### Create a pandas Series with a custom index (use string labels like city names, country names, or anything   (non-integer).
<br>
Access elements using:
- A single label
- A list of labels
- A slice of labels

```python
import pandas as pd
```

```python
values = [100, 200, 300, 400]
index = ['delhi', 'mumbai', 'chennai', 'kolkata']
series_of_names = pd.Series(values, index=index)
display(series_of_names)
```

```python
print(series_of_names['delhi'])
```

```python
print(series_of_names[['delhi','chennai']])
```

```python
print(series_of_names['delhi':'chennai'])
```
