### Find the total, mean, max, and min of the scores.

```python
import pandas as pd

scores = pd.Series([45, 82, 67, 90, 55, 73, 88, 40],
                   index=['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hank'])
```

```python
print("total -> ",scores.sum())
print("mean -> ",scores.mean())
print("max-> ",scores.max())
print("min -> ",scores.min())
```

### Find how many students scored above 70.

```python
print("students scored above 70  \n",scores[(scores>70)].count())
```

### Show only the scores of students who scored above 70 (boolean filtering).

```python
print("student scored above 70 (using boolean filter)",scores[((scores>70) ==True)])
```
