## Question 9: Series Arithmetic Operations

### Tasks:
1. Create two Series `s1` and `s2` with the **same index** (use 4-5 items, e.g. product names)
2. Perform **addition, subtraction, multiplication, division** between `s1` and `s2`
3. Now create two Series `s3` and `s4` with **different/partially overlapping indexes**
4. Add `s3` and `s4` — observe what happens at non-matching indexes
5. Repeat the addition using `add()` method with `fill_value=0` to handle the NaN

```python
import pandas as pd
```

```python
# Task 1: Create s1 and s2 with same index
s1 =pd.Series([1,2,3,4,5],index=('a','b','c','d','e'))
s2 =pd.Series([10,20,30,40,50],index=('a','b','c','d','e'))
```

```python
# Task 2: Addition, subtraction, multiplication, division
print(s1+s2)
print(s1-s2)
print(s1*s2)
print(s1/s2)
```

```python
# Task 3: Create s3 and s4 with different/partially overlapping indexes
s3=pd.Series([1,2,3],index=('a','b','c'))
s4=pd.Series([11,22,33],index=('a','e','d'))
```

```python
# Task 4: Add s3 + s4 using + operator — observe NaN at non-matching indexes
print(s3+s4)
```

```python
# Task 5: Add s3 and s4 using .add() with fill_value=0 to avoid NaN
print(s3.add(s4, fill_value=0))
```
