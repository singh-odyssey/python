- Create a Series with a custom index
- Access elements using loc (label-based) and iloc (position-based)
- Show the difference between them

```python
import pandas as pd
```

```python
values=[100,200,300,400,500]
index= ['delhi', 'mumbai', 'chennai', 'kolkata' ,'banglore']
series =pd.Series(values,index=index)
display(series)
```

```python
print("using loc",series.loc['delhi'])
print("using iloc",series.iloc[0])
```

```python
print("\nloc  slice (inclusive both ends):\n", series.loc['delhi':'chennai'])
print("\niloc slice (exclusive end):\n", series.iloc[0:3])
```
