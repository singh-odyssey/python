# Boolean Masking & Advanced Filtering

### Tasks:
1. Filter students who scored **above 70**
2. Filter students who scored **between 60 and 85** (inclusive)
3. Filter students whose names **start with a vowel** (A, E, I)
4. Use `.where()` to keep scores above 70, replace the rest with `0`

```python
import pandas as pd
```

```python
scores = pd.Series([45, 82, 67, 90, 55, 73, 88, 41, 95, 60],
                   index=['Alice','Bob','Charlie','David','Eva',
                          'Frank','Grace','Hank','Ivy','Jack'])
```

```python

# More idiomatic (no need for .values):
# print(scores[scores > 70])
```

```python
# Task 2: Filter students who scored between 60 and 85 (inclusive)
print(scores[(scores.values >= 60) & (scores.values <= 85)])

# More idiomatic using .between():
# print(scores[scores.between(60, 85)])
```

```python
# Task 3: Filter students whose names start with a vowel (A, E, I)
print(scores[scores.index.str.startswith(('A', 'E', 'I'))])
```

```python
# Task 4: Use .where() to keep scores above 70, replace the rest with 0
print(scores.where(scores > 70, 0))
```
