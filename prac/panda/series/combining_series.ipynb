# Combining Series (concat, align)

### Tasks:
1. Concatenate `month1` and `month2` into a single Series (keep all entries)
2. Add the two Series — matching indexes sum up, missing ones become `NaN`
3. Add them together but **fill missing values with `0`** instead of `NaN`
4. Find names (indexes) that appear in **both** Series

```python
import pandas as pd
```

```python
month1 = pd.Series([200, 350, 150], index=['Alice', 'Bob', 'Charlie'])
month2 = pd.Series([400, 180, 300], index=['Bob', 'Charlie', 'David'])
```

```python
# Task 1: Concatenate month1 and month2 into a single Series
concat_month = pd.concat([month1, month2])
print(concat_month)
```

```python
# Task 2: Add the two Series (missing indexes become NaN)
print(month1+month2)
```

```python
# Task 3: Add them but fill missing values with 0 instead of NaN
print((month1+month2).fillna(0))
```

```python
# Task 4: Find names (indexes) that appear in both Series
print(month1.index.intersection(month2.index))
```
