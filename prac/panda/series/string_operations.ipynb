## Question 10: String Operations on Series (`.str` accessor)

### Tasks:
1. Create a Series of 5 full names (mixed casing, e.g. `"alice smith"`, `"BOB JONES"`)
2. Convert all names to **lowercase**, then **uppercase**, then **title case**
3. Check which names **start with the letter 'a'** (case-insensitive)
4. **Split** each name into first and last name using `.str.split()`
5. **Strip** extra whitespace from `pd.Series(["  hello ", " world ", "  pandas  "])`
6. **Replace** a specific word in the names (e.g. replace `"smith"` with `"kumar"`)
7. Check which names **contain** the substring `"son"` (e.g. `"johnson"`, `"anderson"`)

```python
import pandas as pd
```

```python
# Task 1: Create a Series of 5 full names with mixed casing
names = pd.Series(["alice smith", "BOB JONES", "Alex ANDERSON", "JOHN doe", "diana johnson"])
display(names)
```

```python
# Task 2: Convert to lowercase, uppercase, title case
print("lowercase:\n", names.str.lower())
print("\nuppercase:\n", names.str.upper())
print("\ntitle case:\n", names.str.title())
```

```python
# Task 3: Check which names start with the letter 'a' (case-insensitive)
# convert to lowercase first, then check startswith
print("Names starting with 'a':\n", names[names.str.lower().str.startswith('a')])
```

```python
# Task 4: Split each name into first and last name
print("Split names:\n", names.str.split())
```

```python
# Task 5: Strip extra whitespace
spaces = pd.Series(["  hello ", " world ", "  pandas  "])
print("before strip:\n", spaces)
print("\nafter strip:\n", spaces.str.strip())
```

```python
# Task 6: Replace a specific word (e.g. 'smith' -> 'kumar')  [case-insensitive using regex]
print("after replace:\n", names.str.replace("smith", "kumar", case=False, regex=False))
```

```python
# Task 7: Check which names contain the substring 'son' (case-insensitive)
print("Names containing 'son':\n", names[names.str.lower().str.contains('son')])
```
