## Question 8: Series Sorting

### Tasks:
1. Create a Series of 6 students and their marks (use custom string index for student names, assign random marks)
2. Sort the Series by **values** in ascending order
3. Sort the Series by **values** in descending order
4. Sort the Series by **index** (student names) in ascending order
5. Sort the Series by **index** in descending order
6. Find the **top 3** students (highest marks) after sorting

```python
import pandas as pd
```

```python
# Task 1: Create the Series
s=pd.Series([98,89,87,99,90,92],index=('s1','s2','s3','s4','s5','s6'))
```

```python
# Task 2: Sort by values ascending
s.sort_values()
```

```python
# Task 3: Sort by values descending
s.sort_values(ascending=False)
```

```python
# Task 4: Sort by index ascending
s.sort_index()
```

```python
# Task 5: Sort by index descending
s.sort_index(ascending=False)
```

```python
# Task 6: Top 3 students (highest marks)
print("Top three students -> \n",s.sort_values(ascending=False).head(3))
```
