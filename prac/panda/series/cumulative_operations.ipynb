# Cumulative Operations

### Tasks:
1. Calculate the **cumulative sum** of sales across the week
2. Calculate the **cumulative max** (running highest value so far)
3. Calculate the **cumulative product** of sales
4. Calculate the **running mean** — after each day, what is the average so far? *(Hint: use `expanding()`)*

```python
import pandas as pd
```

```python
sales = pd.Series([120, 85, 200, 95, 150, 60, 175],
                  index=['Mon','Tue','Wed','Thu','Fri','Sat','Sun'])
```

```python
# Task 1: Cumulative sum of sales
sales.cumsum()
```

```python
# Task 2: Cumulative max (running highest value so far)
sales.cummax()
```

```python
# Task 3: Cumulative product of sales
sales.cumprod()
```

```python

# Task 4: Running mean after each day (Hint: use expanding())
sales.expanding().mean()
```
