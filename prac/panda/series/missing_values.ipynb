```python
import pandas as pd

s = pd.Series([10, 20, None, 40, None, 60])
```

### Check if there are any missing values in the Series.

```python
if s.isnull().any():
    print("series has null values ")
```

### Find the count of missing values


```python
print("Number of null values \n", s.isnull().sum())
```

### Fill the missing values with the mean of the non-missing values.

```python
s.fillna((s.mean()))
```

### Drop the missing values instead (show as a separate step).

```python
s.dropna()
```
