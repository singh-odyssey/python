## Question 13: `dtype` and `astype()` — Type Conversion

### Tasks:
1. Create a Series of numbers stored as **strings**: `['10', '20', '30', '40', '50']`
2. Check its `dtype` — what is it?
3. Convert it to **integer** using `astype(int)` and verify the dtype changed
4. Convert that integer Series to **float** using `astype(float)`
5. Create a Series `[1, 0, 1, 1, 0]` and convert it to **boolean** dtype
6. Try to convert `['10', 'abc', '30']` to int — what happens? Handle the error using `pd.to_numeric()` with `errors='coerce'`
7. What does `errors='coerce'` do? Write it as a comment

```python
import pandas as pd
```

```python
# Task 1: Create a Series of numbers stored as strings
s =pd.Series(['1', '2', '3', '4', '5'])
```

```python
# Task 2: Check its dtype
print(s.dtype)
```

```python
# Task 3: Convert to integer using astype(int) and verify dtype
s = s.astype(int)
print(s.dtype)
print(s)
```

```python
# Task 4: Convert integer Series to float using astype(float)
s=s.astype(float)
print(s.dtype)
print(s)
```

```python
# Task 5: Create Series [1, 0, 1, 1, 0] and convert to boolean dtype
sb=pd.Series([1, 0, 1, 1, 0])
sb=sb.astype(bool)
print(sb.dtype)
print(sb)
```

```python
# Task 6: Try converting ['10', 'abc', '30'] to int — handle error with pd.to_numeric(errors='coerce')
s = pd.Series(['10', 'abc', '30'])
s = pd.to_numeric(s, errors='coerce')
print(s)
```

```python
# Task 7: What does errors='coerce' do?
# errors='coerce' converts invalid values to NaN instead of raising an error.
# 
```
