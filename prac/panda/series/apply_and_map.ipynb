## Question 11: `apply()` and `map()`

### Tasks:
1. Create a Series of 5 numbers
2. Use `apply()` with a **lambda** to square each number
3. Use `apply()` with a **named function** to categorize numbers as `'even'` or `'odd'`
4. Create a Series of **grades** like `['A', 'B', 'C', 'D', 'F']`
5. Use `map()` with a **dictionary** to convert grades to marks (e.g. `A → 90, B → 80 ...`)
6. Use `map()` with a **lambda** to add 10 to each number from Task 1
7. What is the difference between `apply()` and `map()`? (write in a comment)

```python
import pandas as pd
```

```python
# Task 1: Create a Series of 5 numbers
s=pd.Series([10, 20, 30, 40, 50])
```

```python
# Task 2: Use apply() with a lambda to square each number
print(s.apply(lambda x:x**2))
```

```python
# Task 3: Use apply() with a named function to label 'even' or 'odd'
def even_odd(x):
    
        if x %2==0:
            return 'even'
        else:
            return 'odd'

print(s.apply(even_odd))
```

```python
# Task 4: Create a Series of grades
grades=pd.Series(['A','B','C','D'])
```

```python
# Task 5: Use map() with a dictionary to convert grades to marks
marks_map = {'A': 90, 'B': 80, 'C': 70, 'D': 60, 'F': 0}
print(grades.map(marks_map))
```

```python
# Task 6: Use map() with a lambda to add 10 to each number from Task 1
print(s.map(lambda x:x+10))
```
