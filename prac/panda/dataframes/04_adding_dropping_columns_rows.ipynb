## ❓ Question 3 — Basic-Intermediate

Start with this DataFrame:

```python
df3 = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age':   [25, 30, 35, 28],
    'score': [88, 45, 72, 39]
})
```

Drop all rows where `score < 50`.  
Print the resulting DataFrame (reset the index).

````python
# ✏️ Your Code Here
df3 = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age':   [25, 30, 35, 28],
    'score': [88, 45, 72, 39]
})
df3 = df3[df3['score'] >= 50].reset_index(drop=True)
df3
````

## ❓ Question 4 — Intermediate

Start with this DataFrame:

```python
df4 = pd.DataFrame({
    'nm':  ['Alice', 'Bob', 'Charlie'],
    'ag':  [25, 30, 35],
    'ct':  ['New York', 'London', 'Paris']
})
```

Rename the columns so that:
- `'nm'` → `'name'`
- `'ag'` → `'age'`
- `'ct'` → `'city'`

Print the renamed DataFrame.

````python
# ✏️ Your Code Here
df4 = pd.DataFrame({
    'nm':  ['Alice', 'Bob', 'Charlie'],
    'ag':  [25, 30, 35],
    'ct':  ['New York', 'London', 'Paris']
})
df4=df4.rename(columns={'nm':'name','ag':'age','ct':'city'})
df4
````

## ❓ Question 5 — Intermediate

Start with this DataFrame:

```python
df5 = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Charlie'],
    'age':   [25, 30, 35],
    'score': [88, 92, 85]
})
```

Insert a new column called `'grade'` with values `['A', 'A+', 'A']`  
at **position 2** (between `age` and `score`).

````python
# ✏️ Your Code Here
df5 = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Charlie'],
    'age':   [25, 30, 35],
    'score': [88, 92, 85]
})
df5.insert(2, 'grade', ['A', 'A+', 'A'])
df5
````
