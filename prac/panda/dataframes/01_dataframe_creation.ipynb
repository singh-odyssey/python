## ❓ Question 1 — Basic

Create a DataFrame from the following dictionary:

```python
data = {
    'name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age':   [25, 30, 35, 28],
    'city':  ['New York', 'London', 'Paris', 'Tokyo']
}
```

Store it in a variable called `df`.  
Print the DataFrame.

````python
# ✏️ Your Code Here
data ={
    'name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age':   [25, 30, 35, 28],
    'city':  ['New York', 'London', 'Paris', 'Tokyo']
}
df =pd.DataFrame(data)
print(df)
````

````python
# ✅ Verify Q1
assert list(df.columns) == ['name', 'age', 'city'], "Column names don't match!"
assert df.shape == (4, 3), "Shape should be (4, 3)!"
assert df['name'].tolist() == ['Alice', 'Bob', 'Charlie', 'Diana'], "Names don't match!"
print("✅ Q1 Passed!")
````

## ❓ Question 2 — Basic

Create a DataFrame from this **list of dictionaries**:

```python
records = [
    {'product': 'Apple',  'price': 1.2,  'stock': 100},
    {'product': 'Banana', 'price': 0.5,  'stock': 200},
    {'product': 'Cherry', 'price': 3.0,  'stock': 50},
```

Store it in a variable called `df2`.

````python
# ✏️ Your Code Here
records = [
    {'product': 'Apple',  'price': 1.2,  'stock': 100},
    {'product': 'Banana', 'price': 0.5,  'stock': 200},
    {'product': 'Cherry', 'price': 3.0,  'stock': 50},
]
df2=pd.DataFrame(records)
print(df2)
````

## ❓ Question 3 — Basic-Intermediate

Create a DataFrame from a **2D list** (list of lists) with custom column names:

```python
rows = [
    ['Math',    90, 85],
    ['Science', 78, 92],
    ['English', 88, 76],
    ['History', 70, 80],
]
columns = ['Subject', 'Midterm', 'Final']
```

Store it in `df3` and print it.

````python
# ✏️ Your Code Here
rows = [
    ['Math',    90, 85],
    ['Science', 78, 92],
    ['English', 88, 76],
    ['History', 70, 80],
]
columns = ['Subject', 'Midterm', 'Final']

df3 = pd.DataFrame(rows, columns=columns)
print(df3)
````

````python
# ✅ Verify Q3
assert list(df3.columns) == ['Subject', 'Midterm', 'Final'], "Column names don't match!"
assert df3.shape == (4, 3), "Shape should be (4, 3)!"
assert df3['Subject'].tolist() == ['Math', 'Science', 'English', 'History']
print("✅ Q3 Passed!")
````

## ❓ Question 4 — Intermediate

Create the same DataFrame as Q1, but this time set the **index** to be the `name` column (use it as the row label, not a regular column).

```python
data = {
    'name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age':   [25, 30, 35, 28],
    'city':  ['New York', 'London', 'Paris', 'Tokyo']
}
```

Store it in `df4`.  
The resulting DataFrame should have `name` as the **index**, and only `age` and `city` as columns.

````python
# ✏️ Your Code Here
data ={
    'name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age':   [25, 30, 35, 28],
    'city':  ['New York', 'London', 'Paris', 'Tokyo']
}
df4 = pd.DataFrame(data).set_index('name')
print(df4)
````

````python
# ✅ Verify Q4
assert df4.index.name == 'name', "Index name should be 'name'!"
assert list(df4.columns) == ['age', 'city'], "Columns should be ['age', 'city']!"
assert df4.loc['Alice', 'city'] == 'New York', "Alice's city should be New York!"
print("✅ Q4 Passed!")
````

## ❓ Question 5 — Intermediate

Create a DataFrame using `pd.DataFrame()` with **numpy** to generate data:

- Use `np.arange(1, 6)` as the `ID` column (values 1 to 5)
- Use `np.random.seed(42)` then `np.random.randint(50, 100, size=5)` as the `Score` column
- Use `np.random.choice(['Pass', 'Fail'], size=5)` as the `Result` column  
  *(keep seed 42, continue from the same random state)*

Store it in `df5` and print it.

````python
# ✏️ Your Code Here
np.random.seed(42)

df5 = pd.DataFrame({
    'ID': np.arange(1, 6),
    'Score': np.random.randint(50, 100, size=5),
    'Result': np.random.choice(['Pass', 'Fail'], size=5)
})

print(df5)
````

````python
# ✅ Verify Q5
assert list(df5.columns) == ['ID', 'Score', 'Result'], "Column names should be ['ID', 'Score', 'Result']!"
assert df5.shape == (5, 3), "Shape should be (5, 3)!"
assert df5['ID'].tolist() == [1, 2, 3, 4, 5], "ID should be 1 to 5!"
assert df5['Score'].dtype in [np.int32, np.int64], "Score should be integer type!"
print("✅ Q5 Passed!")
````
