```python
import pandas as pd
import numpy as np
```

```python
# Base DataFrame for this topic
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan'],
    'age': [25, 30, 35, 28, 40],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin'],
    'score': [88, 92, 85, 90, 95]
}
df = pd.DataFrame(data)
df
```

## Question 1 - Shape and Size
Store the number of rows in `n_rows` and number of columns in `n_cols` using `df.shape`.

```python
# Your Code Here
n_rows, n_cols = df.shape

print(n_rows)
print(n_cols)
```

## Question 2 - Columns and Index
Store column names as a Python list in `col_list`, and index values as a list in `idx_list`.

```python
# Your Code Here
col_list = df.columns.tolist()
idx_list = df.index.tolist()

print(col_list)
print(idx_list)
```

## Question 3 - Data Types
Store a dictionary of dtypes in `dtype_map` using `df.dtypes.to_dict()`.

```python
# Your Code Here
dtype_map =df.dtypes.to_dict()
print(dtype_map)
```

## Question 4 - Head and Tail
Store first 2 rows in `top2` and last 2 rows in `bottom2`.

```python
# Your Code Here
top2 =df.head(2)
bottom2=df.tail(2)

print(top2)
print(bottom2)
```

## Question 5 - Describe Numeric Data
Store numeric summary in `summary` using `df.describe()`.
Then store mean of `score` in `mean_score` from `summary`.

```python
# Your Code Here
summary = df.describe()
print(summary)
print()

mean_score = summary.loc['mean', 'score']
print(mean_score)
```
