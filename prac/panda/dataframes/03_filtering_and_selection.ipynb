```python
import pandas as pd
import numpy as np
```

```python
# Base DataFrame for this topic
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan'],
    'age': [25, 30, 35, 28, 40],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin'],
    'score': [88, 92, 85, 90, 95],
    'department': ['Sales', 'HR', 'Tech', 'Sales', 'Tech']
}
df = pd.DataFrame(data)
df
```

## Question 1 - Select Single Column
Select the `name` column and store it in `names`. Then select the `score` column and store it in `scores`.

```python
# Your Code Here
names=df['name'].tolist()
scores=df['score'].tolist()
print(names)
print(scores)
```

## Question 2 - Select Multiple Columns
Select columns `name`, `age`, and `city` as a new DataFrame and store in `selected_df`.

```python
# Your Code Here
selected_df=df[['name','age','city']]
selected_df
```

## Question 3 - Filter Rows with Condition
Filter rows where `age > 28` and store in `filtered_df`.

```python
# Your Code Here
filtered_df =df[df['age']>28]
filtered_df
```

## Question 4 - Use loc and iloc
Using `loc`, get the row where `name == 'Diana'` and store in `diana_row`.  
Using `iloc`, get the 3rd row (index position 2) and store in `third_row`.

```python
# Your Code Here
diana_row = df.loc[df['name'] == 'Diana']
third_row = df.iloc[2]
diana_row
third_row
```

## Question 5 - Multiple Conditions
Filter rows where `department == 'Tech'` AND `score >= 90` and store in `tech_high_score`.

```python
# Your Code Here
tech_high_score =df[(df['department']=='Tech' )& (df['score']>=90)]
tech_high_score
```
