### Building a model to predict the marks of student based on there studying time .


### Creating a random data of student study hours and marks according to it .

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Generate 1000 random study hours between 1.0 and 10.0 hours
study_hours = np.random.uniform(1.0, 10.0, 1000)

# Define a linear relationship: marks = 35 + 6.5 * study_hours + random noise
# Baseline score is 35%, each hour adds roughly 6.5%, plus some variation (noise)
noise = np.random.normal(0, 4.5, 1000) # Mean 0, standard deviation 4.5
marks = 35 + 6.5 * study_hours + noise

# Clip marks between 0 and 100 to keep it realistic
marks = np.clip(marks, 0, 100)

# Round values for clean data
study_hours = np.round(study_hours, 1)
marks = np.round(marks).astype(int)

# Create DataFrame and save to CSV
df = pd.DataFrame({
    'study_time_hours': study_hours,
    'marks_percentage': marks
})

df.to_csv('student_marks.csv', index=False)
print("Successfully generated student_marks.csv with 1000 rows!")
```
